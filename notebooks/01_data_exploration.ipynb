# 01. Exploratory Data Analysis (EDA)
## Urban Flood Risk Prediction — Greater Mumbai Pilot


```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Load processed dataset
df = pd.read_csv('../data/processed/mumbai_flood_features.csv')
print(f'Shape: {df.shape}')
df.head()
```

```python
# Summary statistics
df.describe()
```

```python
# Class imbalance check
prevalence = df['flood_observed'].value_counts(normalize=True)
print('Flood label balance:\n', prevalence)

plt.figure(figsize=(6, 4))
sns.countplot(data=df, x='flood_observed', palette='Blues')
plt.title('Target Distribution (flood_observed)')
plt.xlabel('Flood Observed (0 = No, 1 = Yes)')
plt.show()
```

```python
# Distribution of 24h Rainfall vs Elevation
plt.figure(figsize=(8, 5))
sns.scatterplot(data=df, x='rain_24h_mm', y='elevation_m', hue='flood_observed', alpha=0.6, palette={0: '#3b82f6', 1: '#ef4444'})
plt.title('Rainfall vs Elevation Inundation Boundary')
plt.show()
```
