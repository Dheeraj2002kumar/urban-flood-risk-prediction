# 02. Geospatial Exploration & Pilot Hotspots Analysis
## Mumbai Pilot Monitoring Stations & Mithi River Basin


```python
import json
import pandas as pd
import matplotlib.pyplot as plt

with open('../frontend/public/sample-area.geojson', 'r') as f:
    geojson_data = json.load(f)

print(f'Features in GeoJSON: {len(geojson_data["features"])}')
for feat in geojson_data['features'][:5]:
    print(feat['properties']['location_id'], feat['properties']['location_name'], feat['properties']['elevation_m'])
```

```python
# Plot station GPS coordinates across Mumbai
lats = [f['properties']['center_latitude'] for f in geojson_data['features']]
lons = [f['properties']['center_longitude'] for f in geojson_data['features']]
elevs = [f['properties']['elevation_m'] for f in geojson_data['features']]

plt.figure(figsize=(7, 9))
scatter = plt.scatter(lons, lats, c=elevs, cmap='viridis_r', s=120, edgecolors='black')
plt.colorbar(scatter, label='Elevation (m)')
plt.title('Mumbai Pilot Monitoring Network Topography')
plt.xlabel('Longitude')
plt.ylabel('Latitude')
plt.grid(True, linestyle='--', alpha=0.5)
plt.show()
```
